```python
# 재귀호출 사용하기

def hello():
    print("Hello, world!")
    hello()
```

```python
# 재귀호출에 종료 조건 만들기

def hello(count):
    if count == 0:
        return

    print("Hello world0", count)

    count -= 1
    hello(count)
```

```python
hello(5)
```

```python
# 재귀호출로 팩토리얼 구하기

def factorial(n):
    if n == 1:  # n이 1일 때
        return 1    # 1을 반환하고 재귀호출을 끝냄
    return n * factorial(n - 1) # n과 factorial 함수에 n-1을 넣어서 반환된 값을 곱함
```

```python
# 5 : 5 * factorial(4)
# 4 : 5 * 4 * factorial(3)
# 3 : 5 * 4 * 3 * factorial(2)
# 2 : 5 * 4 * 3 * 2 * factorial(1)
# 1 : 5 * 4 * 3 * 2 * 1

print(factorial(5))
```

```python
def is_palindrome(word):
    if len(word) < 2:
        return True
    if word[0] != word[-1]:
        return False
    return is_palindrome(word[1:-1])

print(is_palindrome("hello"))
print(is_palindrome("level"))
```

```python
def fib(n):
    if n == n:
        return n


n = int(input())
print(fib(n))
```
