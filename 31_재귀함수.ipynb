```python
# 재귀호출 사용하기

def hello():
    print("Hello, world!")
    hello()
```

```python
# 재귀호출에 종료 조건 만들기

def hello(count):
    if count == 0:      # 종료 조건을 만듦  count가 0이면 다시 hello 함수를 호출하지 않고 끝냄
        return

    print("Hello world0", count)

    count -= 1      # count를 1 감소시킨뒤
    hello(count)    # 다시 hello에 넣음
```

```python
hello(5)
```

```python
# 재귀호출로 팩토리얼 구하기

def factorial(n):
    if n == 1:  # n이 1일 때
        return 1    # 1을 반환하고 재귀호출을 끝냄
    return n * factorial(n - 1) # n과 factorial 함수에 n-1을 넣어서 반환된 값을 곱함
```

```python
# 5 : 5 * factorial(4)
# 4 : 5 * 4 * factorial(3)
# 3 : 5 * 4 * 3 * factorial(2)
# 2 : 5 * 4 * 3 * 2 * factorial(1)
# 1 : 5 * 4 * 3 * 2 * 1

print(factorial(5))
```

```python
def is_palindrome(word):
    if len(word) < 2:
        return True
    if word[0] != word[-1]:     # 첫 번째 문자와 마지막 문자가 다를 때 False를 반환
        return False
    return is_palindrome(word[1:-1])    # 현재 문자열의 두 번째 문자부터 뒤이ㅔ서 두 번째 문자까지만 잘라서 다시 넣어줌

print(is_palindrome("hello"))
print(is_palindrome("level"))
```

```python
def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    
    a = 0
    b = 1

    for i in range(2, n + 1):
        a, b = b, a + b

    return b

n = int(input())
print(fib(n))
```
