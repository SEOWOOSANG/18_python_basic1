```python
# 위치 인수를 사용하는 함수 만들고 호출하기

def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)

print_numbers(10, 20, 30)
```

```python
# 언패킹 사용하기

x = [10, 20, 30]
print_numbers(x) # => ([10, 20, 30])
```

```python
x = [10, 20, 30]
print_numbers(*x)
```

```python
# 가변 인수 함수 만들기
# def 함수이름(*매개변수):
#     코드

def print_numbers(*args):
    for arg in args:
        print(arg)
```

```python
print_numbers(10)
print("=====")
print_numbers(10, 20, 30, 40)
```

```python
# 키워드 인수 사용하기

def personal_info(name, age, address):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
personal_info("홍길동", 30, "서울시 용산구")
```

```python
# 키워드 인수
# 함수(키워드=값)

personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
personal_info(age=30, address="서울시 용산구", name="홍길동")
```

```python
# 키워드 인수와 딕셔너리 언패킹 사용하기
# 함수(**딕션너리)

def personal_info(name, age, address):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)

x = {
    "name": "홍길동",
    "age": 30,
    "address": "서울시 용산구" 
}

personal_info(**x)
# personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
# 키워드 인수를 사용하는 가변 인수 함수 만들기
# def 함수이름(**매개변수):
#     코드

def personal_info(**kwargs):
    print(kwargs)
    print(type(kwargs))
    for kw, arg in kwargs.items():
        print(kw, ": ", arg, sep="")

personal_info(name="홍길동")
```

```python
personal_info(name="홍길동", age=30, address="서울시 용산구")
```

```python
def personal_info(**kwargs):
    for kw, arg in kwargs.items():
        print(kw, ": ", arg, sep="")

x = {
    "name": "홍길동",
    "age": 30,
    "address": "서울시 용산구",
    "hobby": ["축구", "농구", "게임"]
}

personal_info(**x)
```

```python
# 매개변수에 초기값 지정하기
# def 함수이름(매개변수=값):
#     코드

def personal_info(name, age, address="비공개"):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)

personal_info("홍길동", 30)
```

```python
personal_info("홍길동", 30, "서울시 이촌동")
```

```python
# 초기값이 지정된 매개변수의 위치

def personal_info(name, address="비공개", age):
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
def personal_info(name, address="비공개", age=0):   # 디폴트값을 설정한 뒤에는 모두 디폴트값을 설정해줘야한다.
    print("이름: ", name)
    print("나이: ", age)
    print("주소: ", address)
```

```python
korean, english, mathematics, science = 100, 86, 81, 91

def get_max_score(*args):
    return max(args)

max_score = get_max_score(korean, english, mathematics, science)
print("높은 점수:", max_score)

max_score = get_max_score(english, science)
print("높은 점수:", max_score)
```
