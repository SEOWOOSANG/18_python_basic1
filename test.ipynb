```python
url = input("URL을 입력하세요: ")   # https://www.example.com/login

parts = url.split("://")        # url을 ://로 구분
protocol = parts[0]             # protocol = https

parts = parts[1].split("/")     # www.example.com/login을 /로 구분     
domain = parts[0]               # domain = www.example
path = parts[1]                 # path = login

print("프로토콜:", protocol)
print("도메인:", domain)
print("경로:", path)
```

```python
keys = input().split()                      # id password name age
values = input().split()                    # admin 1234 홍길동 30

values[3] = int(values[3])                  # values를 int로 만들기

member = dict(zip(keys, values))            # keys, values를 zip에 넣고 다시 dict에 넣기

result = {}                                 # 빈 리스트

for key, value in member.copy().items():    # member.copy(): 딕셔너리를 복사
                                            # .items(): 딕셔너리의 키와 값을 (kye, value) 형태로 가져온다
    if key != "password":                   # 만약 key 값이 password가 아니라면
        result[key] = value                 # 키와 밸류 값을 출력

print(result)
```

```python
library = {
    "문학": [
        {"제목": "1984", "저자": "조지 오웰", "재고": 3},
        {"제목": "위대한 개츠비", "저자": "F. 스콧 피츠제럴드", "재고": 0}
    ],
    "과학": [
        {"제목": "코스모스", "저자": "칼 세이건", "재고": 5},
        {"제목": "이기적 유전자", "저자": "리처드 도킨스", "재고": 2}
    ]
}

for genre, books in library.items():
    for book in books:
        if book["재고"] == 0:
            print("제목:", book["제목"], "분야:", genre)
```

```python
total = 0

for books in library.values():
    for book in books:
        total += book["재고"]

print("전체 재고 수량:", total)
```

```python
number_a, number_b = map(int, input().split())

for i in range(number_a, number_b + 1):
    if i % 20 == 0:
        print("FourFive", end=" ")
    
    elif i % 4 == 0:
        print("Four", end=" ")
    
    elif i % 5 == 0:
        print("Five", end=" ")
   
    else:
        print(i, end=" ")
```

```python
analyze_scores = list(map(int, input().split()))

def get_average(*args):
    return sum(args) / len(args)    # 평균 구하기

def count_average(*args):
    average = get_average(*args)
    return sum(score >= average for score in args)  # 평균 이상인 숫자 개수 카운팅

print("평균점수: {:.1f}".format(get_average(*analyze_scores)),end="")
print("평균이상인 수의 개수:", count_average(*analyze_scores))
```

```python
inventory = {
    "노트북": 1500000,
    "마우스": 30000,
    "키보드": 50000,
    "모니터": 300000
}

def checkout(**kwargs):
    total = 0

    for product, quantity in kwargs.items():
        if product in inventory:
            price = inventory[product] * quantity
            total += price
            print(f"{product} {quantity}개: {price:,}원")
        else:
            print(f"판매하지 않는 상품입니다: {product}")

    print(f"총 결제 금액: {total:,}원")

checkout(노트북=1, 마우스=2)
print()
checkout(모니터=2, 패드=1)
```
