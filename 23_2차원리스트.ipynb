```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기

a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for x, y in a:      # 리스트의 가로 한 줄에서 요소 두 개를 꺼냄
    print(x, y, end=" ")
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):   # 반복할 때 리스트의 크기 활용
    x, y = a[i]     # 요소 두 개를 한꺼번에 가져오기
    print(x, y)
    i += 1          # 인덱스를 1 증가시킴
```

```python
# while 반복문을 두 번 사용하기

a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=" ")
        j += 1
    print()
    i += 1
```

```python
col, row = map(int, input().split())

mine = []
for _ in range(row):
    mine.append(list(input()))

for i in range(row):
    for j in range(col):
        if mine[i][j] == '.':
            count = 0

            for di in range(-1, 2):
                for dj in range(-1, 2):
                    ni = i + di
                    nj = j + dj

                    if 0 <= ni < row and 0 <= nj < col:
                        if mine[ni][nj] == '*':
                            count += 1

            mine[i][j] = str(count)

for line in mine:
    print(''.join(line))
```

```python
# 가로 크기(col)와 세로 크기(row)를 입력받습니다.
col, row = map(int, input().split())

# 지뢰 정보를 저장할 빈 리스트를 만듭니다.
mine = []

# 입력받은 줄을 row만큼 반복합니다.
for _ in range(row):

    # 입력받은 문자열을 문자 하나씩 리스트에 저장합니다.
    # 예: "..*.." → ['.', '.', '*', '.', '.']
    mine.append(list(input()))

# 행을 0부터 row - 1까지 반복합니다.
for i in range(row):

    # 열을 0부터 col - 1까지 반복합니다.
    for j in range(col):

        ## 현재 칸이 지뢰가 아닌 '.'인지 확인합니다.
        ## 지뢰('*')라면 숫자를 계산할 필요가 없습니다.
        if mine[i][j] == '.':

            # 주변 지뢰의 개수를 저장할 변수를 0으로 만듭니다.
            count = 0

            ## 현재 칸을 기준으로 위, 현재, 아래 행을 확인합니다.
            ## -1 = 위, 0 = 현재, 1 = 아래
            for di in range(-1, 2):

                ## 현재 칸을 기준으로 왼쪽, 현재, 오른쪽 열을 확인합니다.
                ## -1 = 왼쪽, 0 = 현재, 1 = 오른쪽
                for dj in range(-1, 2):

                    ## 검사할 위치의 행 번호를 계산합니다.
                    ni = i + di

                    ## 검사할 위치의 열 번호를 계산합니다.
                    nj = j + dj

                    ## 리스트의 범위를 벗어나지 않는지 확인합니다.
                    ## 예를 들어 맨 위쪽에서 위쪽 칸을 확인하면 -1이 되므로
                    ## 이런 경우에는 검사하지 않습니다.
                    if 0 <= ni < row and 0 <= nj < col:

                        ## 검사하는 위치에 지뢰('*')가 있는지 확인합니다.
                        if mine[ni][nj] == '*':

                            ## 지뢰를 발견하면 개수를 1 증가시킵니다.
                            count += 1

            ## 주변 지뢰 개수를 현재 '.' 위치에 저장합니다.
            ## 숫자를 문자열로 바꾸는 이유는 mine 리스트가 문자로 구성되어 있기 때문입니다.
            mine[i][j] = str(count)

# 결과를 한 줄씩 출력합니다.
for line in mine:

    # 리스트의 문자들을 하나의 문자열로 합쳐서 출력합니다.
    # 예: ['0', '1', '*', '2'] → "01*2"
    print(''.join(line))
```
