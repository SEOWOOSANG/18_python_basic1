```python
# 리스트에 요소 하나 추가하기

a = [10, 20, 30]
a.append(500)
a
len(a)
```

```python
a = []
a.append(10)
print(a)
```

```python
# 리스트 안에 리스트 추가하기

a = [10, 20, 30]
a.append([500, 600])
print(a)
print(len(a))
```

```python
# 리스트 확장하기

a = [10, 20, 30]
a.extend([500, 600])
print(a)
print(len(a))
```

```python
# 리스트의 특정 인덱스에 요소 추가하기

a = [10, 20, 10]
a.insert(2, 500)
print(a)
print(len(a))
```

```python
a = [10, 20, 30]
a.insert(0, 500)
print(a)
```

```python
a = [10, 20, 30]
a.insert(len(a), 500)
a
```

```python
a = [10, 20, 30]
a.insert(1, [500, 600])
a
```

```python
a = [10, 20, 30]
a[1:1] = [500, 600]
a
```

```python
a = [10, 20, 30]
a[1:2] = [500, 600]
a
```

```python
# 리스트에서 특정 인덱스의 요소를 삭제하기

a = [10, 20, 30]
print(a.pop())
print(a)
```

```python
a = [10, 20, 30]
print(a.pop(1))
print(a)
```

```python
a = [10, 20, 30]
del a[1]
print(a)
```

```python
# 리스트에서 특정 값을 찾아서 삭제하기

a = [10, 20, 30]
a.remove(20)
print(a)
```

```python
a = [10, 20, 30, 20]
a.remove(20)
a
```
