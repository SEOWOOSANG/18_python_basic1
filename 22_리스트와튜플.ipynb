```python
# While 반복문으로 요소 출력하기

a = [38, 21, 53, 62, 19]
i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 리스트 표현식 사용하기(리스트 컴프리헨션)
## 1. [식 for 변수 in 리스트]
## 2. list(식 for 변수 in 리스트)
```

```python
a = []
for i in range(10):
    a.append(i)
print(a)
```

```python
a = [i for i in range(10)] # 위에 식을 요약
print(a)
```

```python
b = list([i for i in range(10)])
print(b)
```

```python
print([i for i in range(10)])
print([i + 5 for i in range(10)])
print([i * 2 for i in range(10)])
```

```python
# 리스트 표현식에서 if 조건문 사용하기

a = []
for i in range(10):
    if i % 2 == 0:
        a.append(i)
print(a)
```

```python
a = [i for i in range(10) if i % 2 == 0]  # 위에 식을 요약

print(a)
```

```python
b = [i + 5 for i in range(10) if i % 2 == 1]

print(b)
```

```python
b = []
for i in range(10):
    if i % 2 == 1:
        b.append(i + 5)
print(b)
```

```python
# for 반복문과 if 조건문을 여러 번 사용하기

[i * j for i in range(2, 10) for j in range(1, 10)]
```

```python
# 리스트에서 map 사용하기
# list(map(함수, 리스트))
# tuple(map(함수, 튜플))

a = [1.2, 2.5, 3.7, 4.6]
for i in range(len(a)):
    a[i] = int(a[i])

print(a)
```

```python
a = [1.2, 2.5, 3.7, 4.6]
a = list(map(int, a))

print(a)
```
