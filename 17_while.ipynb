```python
# While 반복문 사용하기

# 초기식
# while 조건식:
#     반복할 코드
#     변화식

i = 0
while i < 10:
    print("Hello world")
    i += 1 # => i = i + 1
```

```python
# 초기값을 1부터 시작하기

i = 1
while i <= 5:
    print("Hello world", i)
    i += 1
```

```python
# 초기값을 감소시키기

i = 5
while i > 0:
    print("Hello", i)
    i -= 1 # => i = i - 1
```

```python
count = int(input())

i = 0
while i < count:        # i가 count보다 작을 때 반복
    print("Hello", i)
    i += 1
```

```python
# 무한 루프

i = 1
while i <= 5:
    print("Hello world", i)
    # i += 1
```
