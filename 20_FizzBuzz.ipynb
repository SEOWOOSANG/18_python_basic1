```python
# 1 ~ 100까지 숫자 출력하기

for i in range(1, 10+1):
    print(i)
```

```python
# 3의 배수일때 Fizz, 5의 배수일때 Buzz

for i in range(1, 20+1):
    if i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 20+1):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 코드 단축하기

for i in range(1, 30+1):
    print("Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or i) 
```

```python
for i in range(1, 100+1):
    if i % 2 == 0 and i % 11 == 0:
        print("FizzBuzz")
    elif i % 2 == 0 :
        print("Fizz")
    elif i % 11 == 0:
        print("Buzz")
    else:
        print(i)
```
