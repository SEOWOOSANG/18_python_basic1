```python
# 딕셔너리 조작하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}
x.setdefault("e")
print(x)
```

```python
x.setdefault("f", 100)

print(x)
```

```python
# 딕셔너리에서 키의 값 수정하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}
x.update(a=90)
print(x)
```

```python
x.update(e=50)
print(x)
```

```python
x.update(a=900, f=60)
print(x)
```

```python
y = ({
    1: "one",
    2: "two"
})

y.update({
    1: "ONE",
    2: "TWO"
})
print(y)
```

```python
y.update([
    [2, "TWO"],
    [4, "FOUR"]
])

print(y)
```

```python
y.update(
    zip([1, 2],["one", "two"])
)

print(y)
```

```python
# 딕셔너리에서 키-값 쌍 삭제하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

print(x.pop("a"))
print(x)
```

```python
x.pop("z", 0)
```

```python
del x["b"]
print(x)
```

```python
# 딕셔너리에서 임의의 키-값 쌍 삭제하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

print(x.popitem())  # 딕셔너리의 마지막 요소 삭제
print(x)
```

```python
# 딕셔너리의 모든 키-값 쌍을 삭제하기

print(x)
x.clear()
print(x)
```

```python
# 딕셔너리에서 키의 값을 가져오기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

print(x["a"])
# print(x["z"]) => 없을 경우 에러 발생

print(x.get("a"))
print(x.get("z", 0))
```

```python
# 딕셔너리에서 키-값 쌍을 모두 가져오기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

print(type(x.items()))
print(x.items())
print(list(x.items()))
```

```python
# keys() => 모든 키를 가져옵니다

x.keys()
```

```python
# values() = 모든 값을 가져옵니다

x.values()
```

```python
# 리스트와 튜플로 딕션너리 만들기

keys = ["a", "b", "c", "d"]
x = dict.fromkeys(keys)
x
```

```python
y = dict.fromkeys(keys, 100)
y
```

```python
# 반복문으로 딕셔너리의 키-값 쌍을 모두 출력하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

for i in x:
    print(i)
```

```python
x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

for key, value in x.items():
    print(key, value)
```

```python
# 딕셔너리의 키만 출력하기

for i in x.keys():
    print(i)
```

```python
# 딕셔너리의 값만 출력하기

for i in x.values():
    print(i)
```
