```python
# 딕셔너리 조작하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}
x.setdefault("e")
print(x)
```

```python
x.setdefault("f", 100)

print(x)
```

```python
# 딕셔너리에서 키의 값 수정하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}
x.update(a=90)
print(x)
```

```python
x.update(e=50)
print(x)
```

```python
x.update(a=900, f=60)
print(x)
```

```python
y = ({
    1: "one",
    2: "two"
})

y.update({
    1: "ONE",
    2: "TWO"
})
print(y)
```

```python
y.update([
    [2, "TWO"],
    [4, "FOUR"]
])

print(y)
```

```python
y.update(
    zip([1, 2],["one", "two"])
)

print(y)
```

```python
# 딕셔너리에서 임의의 키-값 쌍 삭제하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

print(x.popitem())  # 딕셔너리의 마지막 요소 삭제
print(x)
```

```python
# 딕셔너리에서 키의 값을 가져오기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

print(x["a"])
# print(x["z"]) => 없을 경우 에러 발생

print(x.get("a"))
print(x.get("z", 0))
```

```python
# 딕셔너리 표현식 사용하기
# {키:값 for 키, 값 in 딕셔너리}

keys = ["a", "b", "c", "d"]
imsi = dict.fromkeys(keys)
print(imsi)

x = {key:value for key, value in imsi.items()}
print(x)
```

```python
dict.fromkeys(["a", "b", "c", "d"]).keys()
```

```python
x = {key: 0 for key in dict.fromkeys(["a", "b", "c", "d"]).keys()}
x
```

```python
# 딕셔너리 표현식에서 if 조건문 사용하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

for key, value in x.items():
    if value == 20:
        del x[key]

print(x)
```

```python
x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

y = x.copy()

for key, value in x.items():
    if value == 20:
        del x[key]

print(x)
```

```python
x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

x = {key: value for key, value in x.items() if value != 20}
print(x)
```
