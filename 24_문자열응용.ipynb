```python
# 문자열 서식 지정자와 포매팅 사용하기
```

```python
## 서식 지정자로 문자열 넣기
# %s % "문자열"

"I am %s" % "james"
```

```python
name = "maria"
"I am %s" % name
```

```python
# 서식 지정자로 숫자 넣기
# %d % 숫자

"I am %d" % 20

b = 10
"I am %d" % b
```

```python
# 서식 지정자로 소수점 표현하기
# "%f" % 숫자(실수)

"%f" % 2.3 # 기본적으로 소수점 이하 6자리까지 표시
```

```python
# "%.자리수f" %숫자
print("%.2f" % 2.3)        # .2f - 소수점 2자리까지 표시
print("%.5f" % 2.3)        # .5f - 소수점 5자리까지 표시
```

```python
# 서식 지정자로 문자열 정렬하기
# %길이s

"%10s" % "python"
```

```python
# 왼쪽 정렬하기

"%-10s" % "python"
```

```python
# 서식 지정자로 문자열 안에 값 여러 개 넣기
# "%d %s" % (숫자, 문자열)

"Today is %d %s" % (3, "April")
```

```python
# format 메서드 사용하기 - "{인덱스}".format(값)
print("Hello, {0}".format("world"))
print("Hello, {0}".format(100))
```

```python
# format 메서드로 같은 값을 여러 개 넣기
print("Hello, {0} {2} {1}".format("Python", "Script", 3.6))
```

```python
# format 메서드에 인덱스 생략하기 - 인덱스를 생략할 경우 지정한 순서대로 값을 출력
print("Hello {} {} {}".format("python", "Script", 3.6))
```

```python
# format 메서드에서 인덱스 대신 이름 지정하기

print("Hello {language} {version}".format(language="Python", version=3.6))
```

```python
# 문자열 포매팅에 변수를 그대로 사용하기

language = "Python" # 변수값 지정
version = 3.6       # 변수값 지정
print(f"Hello, {language} {version}")   # f = format
```

```python
# format 메서드로 문자열 정렬하기
## "{인덱스:<길이}".format("python")

print("{0:<10}".format("python"))
print("{0:>10}".format("python"))
print("{:>10}".format("python"))
```

```python
a = "python" # 인덱스 값

print(f"{a:<10}")
```

```python
# 숫자 개수 맞추기
## "%0개수d" % 숫자
## "{인덱스:0개수d}".format(숫자)

print("%03d" % 1)
print("{0:03d}".format(35))
```

```python
print("%08.2f" % 3.6)
print("{:08.2f}".format(150.37))
```

```python
# 채우기와 정렬을 조합해서 사용하기

print("{:0<10}".format(15))
print("{:0>10}".format(15))
print("{0:x>10}".format(15))
```

```python
number = 3.6
print(f"{number:10.2f}")
print(f"{number:010.2f}")
```

```python
number = 3000000
print(f"{number:,.2f}")
```

```python
lis = map(int, input().split(";"))
lis = list(lis)
lis = sorted(lis,reverse=True)
for i in lis:
    print(f"{i:9,}")
```

```python
lis = map(int, input().split(";"))

for i in sorted(lis, reverse=True):
    print(f"{i:9,}")
```

```python
lis = list(map(int, input().split(";")))
lis.sort(reverse=True)
for i in lis:
    print(f"{i:9,}")
```
