```python
# 반복문 끝내기

i = 0
while True:
    print(i)
    i += 1
    if i == 10:
        break
```

```python
# for에서 break로 반복문 끝내기

for i in range(10000):
    print(i)
    if i == 10:
        break
```

```python
# continue 코드 실행 건너뛰기

for i in range(10):
    if i % 2 == 0:
        continue
    print(i)        # i가 짝수면 continue로 실행하여 print를 건너뛰고, 홀수면 print로 숫자를 출력
```

```python
# while 반복문에서 continue로 코드 실행 건너뛰기

i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
```

```python
# 입력한 횟수대로 반복하기

count = int(input())

i = 0
while True:
    print(i)
    i += 1
    if i == count:
        break
```

```python
count = int(input())

for i in range(count + 1):
    if i % 2 == 0:
        continue
    print(i)
```

```python
i = 0
while True:
    if i % 10 != 3:
        i += 1
        continue
    if i > 73:
        break
    print(i, end=" ")
    i += 1
```

```python
start, stop = map(int,input().split())

i = start

while True:
    if i % 10 == 3:
        i += 1
        continue
    if i > stop:
        break

    print(i, end=" ")
    i += 1
```
