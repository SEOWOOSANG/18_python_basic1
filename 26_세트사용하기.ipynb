```python
# 집합 연산 사용하기
## 합집합
## 세트1 | 세트2
## set.union(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)
print(set.union(a, b))
print(a.union(b))
```

```python
# 교집합
## a & b
## set.intersection(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a & b)
print(set.intersection(a, b))
print(a.intersection(b))
```

```python
# 차집합
## a - b
## set.difference(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(set.difference(a, b))
print(a.difference(b))
print(b - a)
```

```python
# 대칭차집합
## 세트1 ^ 세트2
## set.symmetric_difference(세트1, 세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a ^ b)
print(set.symmetric_difference(a, b))
print(a.symmetric_difference(b))
```

```python
# 집합 연산 후 할당 연산자 사용하기
## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
a |= {5}
print(a)

a = {1, 2, 3, 4}
a.update({5})
print(a)
```

```python
# intersection_update
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}   #a.symmetric_difference_update({3, 4, 5, 6})
print(a)
```

```python
# 부분집합과 상위집합 확인하기
## 현재세트 <= 다른세트
## 현재세트.issubest(다른세트)

a = {1, 2, 3, 4}
a <= {1, 2, 3, 4}
```

```python
a = {1, 2, 3, 4}
a.issubset({1, 2, 3, 4, 5})
```

```python
a = {1, 2, 3, 4}
a < {1, 2, 3, 4}
```

```python
# 세트가 다른지 같은지 확인하기

a = {1, 2, 3, 4}
print(a == {1, 2, 3, 4})
print(a == {4, 2, 1, 3})
```

```python
# 세트 조작하기
## 세트에 요소 추가하기

a = {1, 2, 3 ,4}
a.add(5)
print(a)
```

```python
## 세트에 요소 삭제하기

a = {1, 2, 3 ,4}
a.remove(3)
print(a)
```

```python
a.discard(a)
print(a)
```

```python
a.discard(3)
print(a)    # 요소가 없을 경우에도 오류가 나지 않음
```

```python
# 세트에서 임의의 요소를 삭제하기

a = {1, 2, 3, 4}
print(a.pop())
print(a)
```

```python
# 세트의 모든 요소 삭제하기
a = {1, 2, 3, 4}
a.clear()
print(a)
```

```python
# 반복문으로 세트의 요소를 모두 출력하기
# for 변수 in 세트:
#     반복할 코드

a = {1, 2, 3 ,4}
for i in a:
    print(i)
```

```python
# 세트 표현식 사용하기
## {식 for 변수 in 반복가능한개체}
## set(식 for 변수 in 반복가능한객체)

a = {i for i in "apple"}
print(a)
```

```python
# 세트 표현식에 if 조건문 사용

a = {i for i in "pineapple" if i not in "apl"}
print(a)
```

```python
a = {i for i in range(1, 101) if i % 3 == 0}
b = {i for i in range(1, 101) if i % 5 == 0}

print(a & b)
```

```python
a, b = map(int,input().split())

a = {i for i in range(1, a+1) if a % i == 0} # a의 약수 출력
b = {i for i in range(1, b+1) if b % i == 0} # b의 약수 출력

divisor = a & b # a, b의 교집합

result = 0
if type(divisor) == set:
    result = sum(divisor)

print(result)
```
