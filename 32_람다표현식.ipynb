```python
# map에 객체를 여러 개 넣기

a = [1, 2, 3, 4, 5]
b = [2, 4, 6, 8, 10]

list(map(lambda x, y: x * y, a, b))
```

```python
# filter 사용하기
# filter(함수, 반복가능객체)

def f(x):
    return x > 5 and x < 10

a = [8, 3, 2, 10 ,15, 7, 1, 9, 0, 11]

# f(8) => True

list(filter(f, a))
```

```python
a = [8, 3, 2, 10 ,15, 7, 1, 9, 0, 11]
list(filter(lambda x: 5 < x < 10, a))
```

```python
# reduce 사용하기

# import random
# from random import randint
# randint(1, 10)

from functools import reduce
# (함수, 반복가능객체)

def f(x, y):
    return x + y

a = [1, 2, 3, 4, 5]
reduce(f, a)
```

```python
a = [1, 2, 3, 4, 5]

from functools import reduce

reduce(lambda x, y: x + y, a)
```
