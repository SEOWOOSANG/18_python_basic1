```python
# 문자열 판단하기
# re.match("패턴", "문자열")

import re
re.match("Hello", "Hello, world!")  # 문자열이 있으므로 정규표현식 매치 객체가 반환됨

re.match("python", "Hello, world!") # 문자열이 없으므로 아무것도 반환되지 않음

```

```python
# 문자열이 맨 앞에 오는지 맨 뒤에 오는지 판단하기
# ^문자열
# 문자열$
# re.search("패턴", "문자열")

re.search("^Hello", "Hello world!") # Hello로 시작하므로 패턴에 매칭됨

re.search("world!$", "Hello world!") # world!로 끝나므로 패턴에 매칭됨
```

```python
# 지정된 문자열이 하나라도 포함되는지 판단하기
# 문자열|문자열
#문자열|문자열|문자열|문자열

re.match("hello|world", "hello") # hello 또는 world가 있으므로 패턴이 매칭됨
```

```python
# 범위 판단하기
# [0-9]*    # 숫자 범위 0부터9까지 중에서 문자(숫자)가 0개 이상 있는지
# [0-9]+    # 숫자 범위 0부터9까지 중에서 문자(숫자)가 1개 이상 있는지

re.match("[0-9]*", "1234")

re.match("[0-9]+", "1234")

re.match("[0-9]+", "abcd")  # abcd는 0부터 9까지 숫자가 1개 이상 없으므로 패턴이 매칭되지 않음
```

```python
print(re.match("a*b", "b"))   # b에는 a가 0개 이상 있으므로 패턴에 매칭 됨
print(re.match("a+b", "b"))   # b에는 a가 1개 이상 없으므로 패턴이 매칭되지 않음
print(re.match("a*b", "abb")) # b에는 a가 0개 이상 있으므로 패턴에 매칭 됨
print(re.match("a+b", "aab")) # b에는 a가 1개 이상 있으므로 패턴에 매칭 됨
```

```python
# 문자가 한 개만 있는지 판단하기
# 문자?      ? 앞의 문자(범위)가 0개 또는 1개인지 판단
# [0-9]?
# .         .이 있는 위치에 아무 문자(숫자)가 1개 있는지 판단

print(re.match("abc?d", "abd"))
print(re.match("abc?d", "abcd"))
print(re.match("abc?d", "abcccd"))
print(re.match("ab[0-9]?c", "ab3c"))
print(re.match("ab.d", "abxd"))
```

```python
# 문자 개수 판단하기
# 문자{개수}
# (문자열){개수}

print(re.match("h{3}", "hhhello"))
print(re.match("h{3}", "hhhhello"))
print(re.match("hello{3}", "hellohellohelloworld"))
print(re.match("(hello){3}", "hellohellohelloworld"))
```

```python
print(re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-1234"))
print(re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-123"))
```

```python
# (문자){시작개수,끝개수}
# (문자열){시작개수,끝개수}
# [0-9]{시작개수,끝개수}

print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{3,4}", "02-123-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{3,4}", "02-1234-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{3,4}", "02-12-1234"))
```

```python
# 숫자와 영문 문자를 조합해서 판단하기
# a-z
# A-Z

print(re.match("[a-zA-Z0-9]+", "Hello1234"))
print(re.match("[A-Z0-9]+", "hello"))
```

```python
# 가-힣

print(re.match("[가-힣]+", "홍길동"))
```

```python
# 특정 문자 범위에 포함되지 않는지 판단하기
# [^범위]*
# [^범위]+

print(re.match("[^A-Z]+", "Hello")) # 대문자를 제외 / 대문자가 있으므로 패턴에 매칭되지 않음
print(re.match("[^A-Z]+", "hello")) # 대문자를 제외 / 대문자가 없으므로 패턴에 매칭됨
```

```python
# ^[범위]*
# ^[범위]+

print(re.search("[A-Z]+", "Hello")) # 대문자로 시작하므로 패턴에 매칭됨
```
